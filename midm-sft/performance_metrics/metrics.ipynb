```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# -------------------------
# 1) Load CSV
# -------------------------
file_path = "./final_metrics.csv"
df = pd.read_csv(file_path)

# Extract metrics
bleu_base  = df.loc[df["Metric"] == "BLEU", "Base"].values[0]
bleu_val   = df.loc[df["Metric"] == "BLEU", "Val"].values[0]

rouge_base = df.loc[df["Metric"] == "ROUGE-L", "Base"].values[0]
rouge_val  = df.loc[df["Metric"] == "ROUGE-L", "Val"].values[0]

meteor_base = df.loc[df["Metric"] == "METEOR", "Base"].values[0]
meteor_val  = df.loc[df["Metric"] == "METEOR", "Val"].values[0]

bert_base = df.loc[df["Metric"] == "BERTScore(F1)", "Base"].values[0]
bert_val  = df.loc[df["Metric"] == "BERTScore(F1)", "Val"].values[0]

emb_base = df.loc[df["Metric"] == "Embedding Cosine", "Base"].values[0]
emb_val  = df.loc[df["Metric"] == "Embedding Cosine", "Val"].values[0]


# ----------------------------------------------------
# 2) BLEU – 두 막대 모두 파란색+주황색
# ----------------------------------------------------
plt.figure(figsize=(5, 4), dpi=300)
x = np.arange(2)

bars1 = plt.bar(x[0], bleu_base, color="#1f77b4", label="Baseline")
bars2 = plt.bar(x[1], bleu_val, color="#ff7f0e", label="Fine-tuned(Val)")

# 숫자 표시
for bar in [bars1, bars2]:
    for b in bar:
        height = b.get_height()
        plt.text(b.get_x() + b.get_width()/2, height + 0.001,
                 f"{height:.4f}", ha='center', va='bottom', fontsize=9)

plt.xticks(x, ["Baseline", "Fine-tuned"])
plt.ylabel("Score")
plt.title("BLEU Comparison")
plt.ylim(0, 0.05)
plt.legend()
plt.tight_layout()
plt.savefig("./plot_bleu.png")
plt.close()



# ----------------------------------------------------
# 3) ROUGE-L / METEOR – 기본 파랑/주황 + y축 0~1
# ----------------------------------------------------
plt.figure(figsize=(5, 4), dpi=300)
x = np.arange(2)
width = 0.35

bars1 = plt.bar(x - width/2, [rouge_base, meteor_base], width,
                label="Baseline", color="#1f77b4")
bars2 = plt.bar(x + width/2, [rouge_val, meteor_val], width,
                label="Fine-tuned(Val)", color="#ff7f0e")

# 숫자 표시
for bars in [bars1, bars2]:
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2, height + 0.015,
                 f"{height:.3f}", ha='center', va='bottom', fontsize=9)

plt.xticks(x, ["ROUGE-L", "METEOR"])
plt.ylabel("Score")
plt.title("Lexical Metrics Comparison")
plt.ylim(0, 1)
plt.legend()
plt.tight_layout()
plt.savefig("./plot_lexical.png")
plt.close()



# ----------------------------------------------------
# 4) BERTScore(F1) / Embedding Cosine – y축 0~1
# ----------------------------------------------------
plt.figure(figsize=(5, 4), dpi=300)
x = np.arange(2)
width = 0.35

bars1 = plt.bar(x - width/2, [bert_base, emb_base], width,
                label="Baseline", color="#1f77b4")
bars2 = plt.bar(x + width/2, [bert_val, emb_val], width,
                label="Fine-tuned(Val)", color="#ff7f0e")

# 숫자 표시
for bars in [bars1, bars2]:
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2, height + 0.015,
                 f"{height:.3f}", ha='center', va='bottom', fontsize=9)

plt.xticks(x, ["BERTScore(F1)", "Embedding Cosine"])
plt.ylabel("Score")
plt.title("Semantic Metrics Comparison")
plt.ylim(0, 1)
plt.legend()
plt.tight_layout()
plt.savefig("./plot_semantic.png")
plt.close()
```
