# 2.데이터 전처리 및 분할/
모델링 데이터셋 준비

```python
import os
import shutil
import random
from pathlib import Path
from typing import List, Tuple
import json

class DatasetPreparation:
    def __init__(self, base_path: str = None):
    # base_path가 없으면 현재 작업 디렉토리 사용
        if base_path is None:
            base_path = os.getcwd()
    
        self.base_path = Path(base_path).resolve()  # 절대 경로로 변환
        self.matched_images_path = self.base_path / "matched_images/matched_images"  # 여기 수정!
        self.dataset_path = self.base_path / "dataset"
        
        # 데이터셋 폴더 생성
        self.train_path = self.dataset_path / "train"
        self.val_path = self.dataset_path / "val"
        self.test_path = self.dataset_path / "test"
        
        print(f"기본 경로: {self.base_path}")
        print(f"matched_images 경로: {self.matched_images_path}")
        
    def scan_data(self) -> List[Tuple[str, str, str]]:
        """
        matched_images 폴더에서 데이터 스캔
        Returns: [(가구있는이미지, 빈방이미지, 텍스트파일), ...]
        """
        data_triplets = []
        
        if not self.matched_images_path.exists():
            raise FileNotFoundError(f"matched_images 폴더를 찾을 수 없습니다: {self.matched_images_path}")
        
        # 모든 하위 폴더 스캔
        for folder in self.matched_images_path.iterdir():
            if not folder.is_dir():
                continue
                
            # 폴더 내 파일들 찾기
            jpg_files = list(folder.glob("*.jpg"))
            txt_files = list(folder.glob("*.txt"))
            
            # bf_로 시작하는 파일(빈방)과 일반 jpg 파일(가구있는방) 구분
            furnished_img = None
            empty_img = None
            txt_file = None
            
            for jpg in jpg_files:
                if jpg.name.startswith("bf_"):
                    empty_img = jpg
                else:
                    furnished_img = jpg
            
            if txt_files:
                txt_file = txt_files[0]
            
            # 세 파일이 모두 있는 경우만 추가
            if furnished_img and empty_img and txt_file:
                data_triplets.append((str(furnished_img), str(empty_img), str(txt_file)))
        
        print(f"총 {len(data_triplets)}개의 데이터 세트를 찾았습니다.")
        return data_triplets
    
    def split_dataset(self, data_triplets: List[Tuple[str, str, str]], 
                     train_ratio: float = 0.8, 
                     val_ratio: float = 0.1,
                     test_ratio: float = 0.1,
                     seed: int = 42):
        """
        데이터셋을 train/val/test로 분할 (8:1:1)
        """
        assert abs(train_ratio + val_ratio + test_ratio - 1.0) < 1e-6, "비율의 합이 1이어야 합니다"
        
        random.seed(seed)
        random.shuffle(data_triplets)
        
        total = len(data_triplets)
        train_size = int(total * train_ratio)
        val_size = int(total * val_ratio)
        
        train_data = data_triplets[:train_size]
        val_data = data_triplets[train_size:train_size + val_size]
        test_data = data_triplets[train_size + val_size:]
        
        print(f"\n데이터 분할 완료:")
        print(f"  Train: {len(train_data)}개")
        print(f"  Val: {len(val_data)}개")
        print(f"  Test: {len(test_data)}개")
        
        return train_data, val_data, test_data
    
    def copy_data_to_split(self, data_list: List[Tuple[str, str, str]], 
                          target_path: Path):
        """
        데이터를 target_path로 복사
        """
        target_path.mkdir(parents=True, exist_ok=True)
        
        metadata = []
        
        for idx, (furnished, empty, txt) in enumerate(data_list):
            # 새 파일명 생성
            furnished_dest = target_path / f"furnished_{idx}.jpg"
            empty_dest = target_path / f"empty_{idx}.jpg"
            txt_dest = target_path / f"caption_{idx}.txt"
            
            # 파일 복사
            shutil.copy2(furnished, furnished_dest)
            shutil.copy2(empty, empty_dest)
            shutil.copy2(txt, txt_dest)
            
            # 메타데이터 저장
            metadata.append({
                "id": idx,
                "furnished_image": str(furnished_dest),
                "empty_image": str(empty_dest),
                "caption_file": str(txt_dest)
            })
        
        # 메타데이터를 JSON으로 저장
        metadata_file = target_path / "metadata.json"
        with open(metadata_file, 'w', encoding='utf-8') as f:
            json.dump(metadata, f, indent=2, ensure_ascii=False)
        
        print(f"{target_path.name}에 {len(data_list)}개 파일 복사 완료")
    
    def prepare(self):
        """
        전체 데이터 준비 프로세스 실행
        """
        print("=" * 60)
        print("데이터 준비 시작")
        print("=" * 60)
        
        # 1. 데이터 스캔
        data_triplets = self.scan_data()
        
        if len(data_triplets) == 0:
            raise ValueError("데이터를 찾을 수 없습니다. matched_images 폴더를 확인하세요.")
        
        # 2. 데이터 분할
        train_data, val_data, test_data = self.split_dataset(data_triplets)
        
        # 3. 데이터 복사
        self.copy_data_to_split(train_data, self.train_path)
        self.copy_data_to_split(val_data, self.val_path)
        self.copy_data_to_split(test_data, self.test_path)
        
        print("\n" + "=" * 60)
        print("데이터 준비 완료!")
        print(f"데이터셋 위치: {self.dataset_path}")
        print("=" * 60)
        
        return {
            "train_size": len(train_data),
            "val_size": len(val_data),
            "test_size": len(test_data),
            "dataset_path": str(self.dataset_path)
        }


if __name__ == "__main__":
    # 데이터 준비 실행
    prep = DatasetPreparation()
    result = prep.prepare()
    
    print("\n데이터 준비 결과:")
    for key, value in result.items():
        print(f"  {key}: {value}")
```

# 3.모델링 데이터 셋 로더 

```python
import torch
from torch.utils.data import Dataset, DataLoader
from PIL import Image
from pathlib import Path
import json
from torchvision import transforms
from typing import Optional, Tuple


class EmptyRoomDataset(Dataset):
    """
    빈방 생성을 위한 데이터셋 클래스
    """
    def __init__(self, 
                 data_path: str,
                 image_size: int = 512,
                 augmentation: bool = False):
        """
        Args:
            data_path: train/val/test 폴더 경로
            image_size: 이미지 크기
            augmentation: 데이터 증강 여부
        """
        self.data_path = Path(data_path)
        self.image_size = image_size
        self.augmentation = augmentation
        
        # 메타데이터 로드
        metadata_file = self.data_path / "metadata.json"
        with open(metadata_file, 'r', encoding='utf-8') as f:
            self.metadata = json.load(f)
        
        # Transform 정의
        self.transform = self._get_transform()
        
    def _get_transform(self):
        """이미지 변환 파이프라인"""
        transform_list = [
            transforms.Resize((self.image_size, self.image_size)),
        ]
        
        if self.augmentation:
            transform_list.extend([
                transforms.RandomHorizontalFlip(p=0.5),
                transforms.ColorJitter(brightness=0.1, contrast=0.1, saturation=0.1),
            ])
        
        transform_list.extend([
            transforms.ToTensor(),
            transforms.Normalize([0.5], [0.5])  # [-1, 1] 범위로 정규화
        ])
        
        return transforms.Compose(transform_list)
    
    def __len__(self):
        return len(self.metadata)
    
    def __getitem__(self, idx) -> Tuple[torch.Tensor, torch.Tensor, str]:
        """
        Returns:
            furnished_image: 가구가 있는 이미지 (입력)
            empty_image: 빈방 이미지 (타겟)
            caption: 방 설명 텍스트
        """
        item = self.metadata[idx]
        
        # 이미지 로드
        furnished_img = Image.open(item['furnished_image']).convert('RGB')
        empty_img = Image.open(item['empty_image']).convert('RGB')
        
        # 캡션 로드
        with open(item['caption_file'], 'r', encoding='utf-8') as f:
            caption = f.read().strip()
        
        # Transform 적용
        furnished_tensor = self.transform(furnished_img)
        empty_tensor = self.transform(empty_img)
        
        return furnished_tensor, empty_tensor, caption
    
    def get_raw_images(self, idx) -> Tuple[Image.Image, Image.Image, str]:
        """원본 이미지 가져오기 (평가용)"""
        item = self.metadata[idx]
        
        furnished_img = Image.open(item['furnished_image']).convert('RGB')
        empty_img = Image.open(item['empty_image']).convert('RGB')
        
        with open(item['caption_file'], 'r', encoding='utf-8') as f:
            caption = f.read().strip()
        
        return furnished_img, empty_img, caption


def create_dataloaders(
    dataset_path: str = None,
    batch_size: int = 4,
    num_workers: int = 2,
    image_size: int = 512
):
    """
    Train/Val/Test 데이터로더 생성
    """
    # dataset_path가 없으면 현재 디렉토리 기준
    if dataset_path is None:
        dataset_path = Path.cwd() / "dataset"
    else:
        dataset_path = Path(dataset_path)
    
    # 데이터셋 생성
    train_dataset = EmptyRoomDataset(
        data_path=str(dataset_path / "train"),
        image_size=image_size,
        augmentation=True
    )
    
    val_dataset = EmptyRoomDataset(
        data_path=str(dataset_path / "val"),
        image_size=image_size,
        augmentation=False
    )
    
    test_dataset = EmptyRoomDataset(
        data_path=str(dataset_path / "test"),
        image_size=image_size,
        augmentation=False
    )
    
    # 데이터로더 생성
    train_loader = DataLoader(
        train_dataset,
        batch_size=batch_size,
        shuffle=True,
        num_workers=num_workers,
        pin_memory=True
    )
    
    val_loader = DataLoader(
        val_dataset,
        batch_size=batch_size,
        shuffle=False,
        num_workers=num_workers,
        pin_memory=True
    )
    
    test_loader = DataLoader(
        test_dataset,
        batch_size=1,
        shuffle=False,
        num_workers=num_workers,
        pin_memory=True
    )
    
    print(f"데이터로더 생성 완료:")
    print(f"  Train: {len(train_dataset)}개 샘플, {len(train_loader)}개 배치")
    print(f"  Val: {len(val_dataset)}개 샘플, {len(val_loader)}개 배치")
    print(f"  Test: {len(test_dataset)}개 샘플, {len(test_loader)}개 배치")
    
    return train_loader, val_loader, test_loader, train_dataset, val_dataset, test_dataset


if __name__ == "__main__":
    # 데이터로더 테스트
    train_loader, val_loader, test_loader, _, _, _ = create_dataloaders(
        batch_size=2,
        num_workers=0  # Windows에서는 0으로 설정
    )
    
    # 샘플 데이터 확인
    print("\n샘플 데이터 로드 테스트:")
    for furnished, empty, caption in train_loader:
        print(f"  Furnished image shape: {furnished.shape}")
        print(f"  Empty image shape: {empty.shape}")
        print(f"  Caption: {caption[0][:100]}...")
        break
    
    print("\n데이터로더 테스트 완료!")
```

```python
import torch
import torch.nn as nn
import os
from diffusers import AutoencoderKL, UNet2DConditionModel, DDPMScheduler
from transformers import CLIPTextModel, CLIPTokenizer
from typing import Optional

# hf_transfer 비활성화 (호환성 문제 방지)
os.environ['HF_HUB_ENABLE_HF_TRANSFER'] = '0'


class EmptyRoomModel(nn.Module):
    """
    가구 제거를 위한 빈방 생성 모델
    Stable Diffusion 기반 Image-to-Image 모델
    """
    def __init__(self, 
                 pretrained_model_name: str = "runwayml/stable-diffusion-v1-5",
                 device: str = "cuda" if torch.cuda.is_available() else "cpu"):
        super().__init__()
        
        self.device = device
        
        print(f"모델 로딩 중... (디바이스: {device})")
        print(f"사용 모델: {pretrained_model_name}")
        
        try:
            # VAE (이미지 인코더/디코더)
            print("VAE 로딩 중...")
            self.vae = AutoencoderKL.from_pretrained(
                pretrained_model_name,
                subfolder="vae"
            ).to(device)
            if device == "cuda":
                self.vae = self.vae.half()  # float16 변환
            
            # UNet (노이즈 예측 모델)
            print("UNet 로딩 중...")
            self.unet = UNet2DConditionModel.from_pretrained(
                pretrained_model_name,
                subfolder="unet"
            ).to(device)
            if device == "cuda":
                self.unet = self.unet.half()  # float16 변환
            
            # Text Encoder
            print("Text Encoder 로딩 중...")
            self.text_encoder = CLIPTextModel.from_pretrained(
                pretrained_model_name,
                subfolder="text_encoder"
            ).to(device)
            if device == "cuda":
                self.text_encoder = self.text_encoder.half()  # float16 변환
            
            # Tokenizer - SD 1.5에서 직접 로드 시도
            print("Tokenizer 로딩 중...")
            try:
                # 먼저 SD 1.5의 tokenizer 시도
                self.tokenizer = CLIPTokenizer.from_pretrained(
                    pretrained_model_name,
                    subfolder="tokenizer"
                )
            except:
                # 실패하면 CLIP 모델에서 직접 로드
                print("  -> 대체 tokenizer 사용 중...")
                from transformers import AutoTokenizer
                self.tokenizer = AutoTokenizer.from_pretrained(
                    "openai/clip-vit-large-patch14",
                    use_fast=False
                )
            
            # Noise Scheduler
            print("Scheduler 로딩 중...")
            self.noise_scheduler = DDPMScheduler.from_pretrained(
                pretrained_model_name,
                subfolder="scheduler"
            )
            
            # VAE scaling factor
            self.vae_scale_factor = 2 ** (len(self.vae.config.block_out_channels) - 1)
            
            # 학습 시에는 text_encoder와 vae를 freeze
            self.vae.requires_grad_(False)
            self.text_encoder.requires_grad_(False)
            
            print("모델 로딩 완료!")
            
        except Exception as e:
            print(f"모델 로딩 중 오류 발생: {e}")
            print("\n해결 방법:")
            print("1. 인터넷 연결 확인")
            print("2. 필수 패키지 설치:")
            print("   pip install --upgrade transformers diffusers accelerate")
            print("3. Hugging Face 토큰 설정 (필요시):")
            print("   huggingface-cli login")
            raise
    
    def encode_text(self, prompts):
        """텍스트를 임베딩으로 변환"""
        text_inputs = self.tokenizer(
            prompts,
            padding="max_length",
            max_length=self.tokenizer.model_max_length,
            truncation=True,
            return_tensors="pt"
        )
        
        text_input_ids = text_inputs.input_ids.to(self.device)
        
        with torch.no_grad():
            text_embeddings = self.text_encoder(text_input_ids)[0]
        
        return text_embeddings
    
    def encode_image(self, images):
        """이미지를 latent space로 인코딩"""
        with torch.no_grad():
            latents = self.vae.encode(images).latent_dist.sample()
            latents = latents * self.vae.config.scaling_factor
        return latents
    
    def decode_latents(self, latents):
        """latent를 이미지로 디코딩"""
        latents = latents / self.vae.config.scaling_factor
        with torch.no_grad():
            images = self.vae.decode(latents).sample
        return images
    
    def forward(self, 
                furnished_images: torch.Tensor,
                empty_images: torch.Tensor,
                captions: list):
        """
        학습을 위한 forward pass
        
        Args:
            furnished_images: 가구가 있는 이미지 [B, 3, H, W]
            empty_images: 빈방 이미지 (타겟) [B, 3, H, W]
            captions: 방 설명 텍스트 리스트
        
        Returns:
            loss: 학습 손실
        """
        batch_size = furnished_images.shape[0]
        
        # 1. 텍스트 인코딩
        text_embeddings = self.encode_text(captions)
        
        # 2. 입력 이미지와 타겟 이미지를 latent로 변환
        furnished_latents = self.encode_image(furnished_images)
        empty_latents = self.encode_image(empty_images)
        
        # 3. 노이즈 추가
        noise = torch.randn_like(empty_latents)
        timesteps = torch.randint(
            0, self.noise_scheduler.config.num_train_timesteps,
            (batch_size,), device=self.device
        ).long()
        
        # 타겟에 노이즈 추가
        noisy_latents = self.noise_scheduler.add_noise(empty_latents, noise, timesteps)
        
        # 4. 입력 latent와 노이즈가 추가된 타겟 latent를 concat
        # (입력 조건 추가)
        latent_model_input = torch.cat([noisy_latents, furnished_latents], dim=1)
        
        # 5. UNet으로 노이즈 예측
        noise_pred = self.unet(
            latent_model_input,
            timesteps,
            encoder_hidden_states=text_embeddings
        ).sample
        
        # 6. Loss 계산 (MSE)
        loss = nn.functional.mse_loss(noise_pred, noise)
        
        return loss
    
    @torch.no_grad()
    def generate(self, 
                 furnished_image: torch.Tensor,
                 caption: str,
                 num_inference_steps: int = 50,
                 guidance_scale: float = 7.5):
        """
        추론: 가구가 있는 이미지에서 빈방 생성
        
        Args:
            furnished_image: 가구가 있는 이미지 [1, 3, H, W]
            caption: 방 설명
            num_inference_steps: 디노이징 스텝 수
            guidance_scale: classifier-free guidance 강도
        
        Returns:
            generated_image: 생성된 빈방 이미지
        """
        self.eval()
        
        # 텍스트 인코딩
        text_embeddings = self.encode_text([caption])
        
        # Unconditional embedding (classifier-free guidance용)
        uncond_embeddings = self.encode_text([""])
        text_embeddings = torch.cat([uncond_embeddings, text_embeddings])
        
        # 입력 이미지 인코딩
        furnished_latents = self.encode_image(furnished_image)
        
        # 랜덤 노이즈로 시작
        latents = torch.randn_like(furnished_latents)
        
        # 스케줄러 설정
        self.noise_scheduler.set_timesteps(num_inference_steps)
        
        # 디노이징 루프
        for t in self.noise_scheduler.timesteps:
            # latent 복제 (classifier-free guidance용)
            latent_model_input = torch.cat([latents] * 2)
            latent_model_input = torch.cat([latent_model_input, 
                                           furnished_latents.repeat(2, 1, 1, 1)], dim=1)
            
            # 노이즈 예측
            noise_pred = self.unet(
                latent_model_input,
                t,
                encoder_hidden_states=text_embeddings
            ).sample
            
            # Classifier-free guidance
            noise_pred_uncond, noise_pred_text = noise_pred.chunk(2)
            noise_pred = noise_pred_uncond + guidance_scale * (noise_pred_text - noise_pred_uncond)
            
            # 다음 스텝의 latent 계산
            latents = self.noise_scheduler.step(noise_pred, t, latents).prev_sample
        
        # Latent를 이미지로 디코딩
        generated_image = self.decode_latents(latents)
        
        return generated_image
    
    def save_model(self, save_path: str):
        """모델 저장"""
        torch.save({
            'unet_state_dict': self.unet.state_dict(),
        }, save_path)
        print(f"모델 저장 완료: {save_path}")
    
    def load_model(self, load_path: str):
        """모델 로드"""
        checkpoint = torch.load(load_path, map_location=self.device, weights_only=True)
        self.unet.load_state_dict(checkpoint['unet_state_dict'])
        print(f"모델 로드 완료: {load_path}")


if __name__ == "__main__":
    # 모델 테스트
    print("=" * 60)
    print("모델 초기화 테스트")
    print("=" * 60)
    
    # GPU 확인
    print(f"\nPyTorch 버전: {torch.__version__}")
    print(f"CUDA 사용 가능: {torch.cuda.is_available()}")
    if torch.cuda.is_available():
        print(f"CUDA 버전: {torch.version.cuda}")
        print(f"GPU: {torch.cuda.get_device_name(0)}")
        print(f"GPU 메모리: {torch.cuda.get_device_properties(0).total_memory / 1e9:.2f} GB")
    
    print("\n" + "=" * 60)
    
    try:
        model = EmptyRoomModel()
        trainable_params = sum(p.numel() for p in model.unet.parameters() if p.requires_grad)
        total_params = sum(p.numel() for p in model.parameters())
        
        print("\n" + "=" * 60)
        print("모델 정보")
        print("=" * 60)
        print(f"학습 가능한 파라미터: {trainable_params:,}")
        print(f"전체 파라미터: {total_params:,}")
        print(f"VAE Scale Factor: {model.vae_scale_factor}")
        print("=" * 60)
        print("\n✓ 모델 초기화 성공!")
        
    except Exception as e:
        print(f"\n❌ 오류 발생: {e}")
        print("\n필요한 패키지 설치:")
        print("pip install --upgrade transformers diffusers accelerate")
```

```python
# ============================================================
# 1. 필요한 패키지 import 및 설정
# ============================================================
import os
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from torch.utils.tensorboard import SummaryWriter
from pathlib import Path
from tqdm import tqdm
import time
from datetime import datetime
from PIL import Image
from torchvision import transforms

from diffusers import AutoencoderKL, UNet2DConditionModel, DDPMScheduler
from transformers import CLIPTextModel, CLIPTokenizer

# hf_transfer 비활성화
os.environ['HF_HUB_ENABLE_HF_TRANSFER'] = '0'

# ============================================================
# 2. EmptyRoomModel 클래스
# ============================================================
class EmptyRoomModel(nn.Module):
    """가구 제거를 위한 빈방 생성 모델"""
    
    def __init__(self, 
                 pretrained_model_name: str = "runwayml/stable-diffusion-v1-5",
                 device: str = "cuda" if torch.cuda.is_available() else "cpu"):
        super().__init__()
        
        self.device = device
        
        print(f"모델 로딩 중... (디바이스: {device})")
        print(f"사용 모델: {pretrained_model_name}")
        
        try:
            # VAE
            print("VAE 로딩 중...")
            self.vae = AutoencoderKL.from_pretrained(
                pretrained_model_name,
                subfolder="vae"
            ).to(device)
            if device == "cuda":
                self.vae = self.vae.half()
            
            # UNet
            print("UNet 로딩 중...")
            self.unet = UNet2DConditionModel.from_pretrained(
                pretrained_model_name,
                subfolder="unet"
            ).to(device)
            if device == "cuda":
                self.unet = self.unet.half()
            
            # Text Encoder
            print("Text Encoder 로딩 중...")
            self.text_encoder = CLIPTextModel.from_pretrained(
                pretrained_model_name,
                subfolder="text_encoder"
            ).to(device)
            if device == "cuda":
                self.text_encoder = self.text_encoder.half()
            
            # Tokenizer
            print("Tokenizer 로딩 중...")
            try:
                self.tokenizer = CLIPTokenizer.from_pretrained(
                    pretrained_model_name,
                    subfolder="tokenizer"
                )
            except:
                print("  -> 대체 tokenizer 사용 중...")
                from transformers import AutoTokenizer
                self.tokenizer = AutoTokenizer.from_pretrained(
                    "openai/clip-vit-large-patch14",
                    use_fast=False
                )
            
            # Noise Scheduler
            print("Scheduler 로딩 중...")
            self.noise_scheduler = DDPMScheduler.from_pretrained(
                pretrained_model_name,
                subfolder="scheduler"
            )
            
            # VAE scaling factor
            self.vae_scale_factor = 2 ** (len(self.vae.config.block_out_channels) - 1)
            
            # Freeze VAE and text encoder
            self.vae.requires_grad_(False)
            self.text_encoder.requires_grad_(False)
            
            print("모델 로딩 완료!")
            
        except Exception as e:
            print(f"모델 로딩 중 오류 발생: {e}")
            raise
    
    def encode_text(self, prompts):
        """텍스트를 임베딩으로 변환"""
        text_inputs = self.tokenizer(
            prompts,
            padding="max_length",
            max_length=self.tokenizer.model_max_length,
            truncation=True,
            return_tensors="pt"
        )
        
        text_input_ids = text_inputs.input_ids.to(self.device)
        
        with torch.no_grad():
            text_embeddings = self.text_encoder(text_input_ids)[0]
        
        return text_embeddings
    
    def encode_image(self, images):
        """이미지를 latent space로 인코딩"""
        with torch.no_grad():
            latents = self.vae.encode(images).latent_dist.sample()
            latents = latents * self.vae.config.scaling_factor
        return latents
    
    def decode_latents(self, latents):
        """latent를 이미지로 디코딩"""
        latents = latents / self.vae.config.scaling_factor
        with torch.no_grad():
            images = self.vae.decode(latents).sample
        return images
    
    def forward(self, 
                furnished_images: torch.Tensor,
                empty_images: torch.Tensor,
                captions: list):
        """학습을 위한 forward pass"""
        batch_size = furnished_images.shape[0]
        
        # 텍스트 인코딩
        text_embeddings = self.encode_text(captions)
        
        # 이미지를 latent로 변환
        furnished_latents = self.encode_image(furnished_images)
        empty_latents = self.encode_image(empty_images)
        
        # 노이즈 추가
        noise = torch.randn_like(empty_latents)
        timesteps = torch.randint(
            0, self.noise_scheduler.config.num_train_timesteps,
            (batch_size,), device=self.device
        ).long()
        
        noisy_latents = self.noise_scheduler.add_noise(empty_latents, noise, timesteps)
        
        # latent concat
        latent_model_input = torch.cat([noisy_latents, furnished_latents], dim=1)
        
        # 노이즈 예측
        noise_pred = self.unet(
            latent_model_input,
            timesteps,
            encoder_hidden_states=text_embeddings
        ).sample
        
        # Loss 계산
        loss = nn.functional.mse_loss(noise_pred, noise)
        
        return loss
    
    @torch.no_grad()
    def generate(self, 
                 furnished_image: torch.Tensor,
                 caption: str,
                 num_inference_steps: int = 50,
                 guidance_scale: float = 7.5):
        """추론: 가구가 있는 이미지에서 빈방 생성"""
        self.eval()
        
        text_embeddings = self.encode_text([caption])
        uncond_embeddings = self.encode_text([""])
        text_embeddings = torch.cat([uncond_embeddings, text_embeddings])
        
        furnished_latents = self.encode_image(furnished_image)
        latents = torch.randn_like(furnished_latents)
        
        self.noise_scheduler.set_timesteps(num_inference_steps)
        
        for t in self.noise_scheduler.timesteps:
            latent_model_input = torch.cat([latents] * 2)
            latent_model_input = torch.cat([latent_model_input, 
                                           furnished_latents.repeat(2, 1, 1, 1)], dim=1)
            
            noise_pred = self.unet(
                latent_model_input,
                t,
                encoder_hidden_states=text_embeddings
            ).sample
            
            noise_pred_uncond, noise_pred_text = noise_pred.chunk(2)
            noise_pred = noise_pred_uncond + guidance_scale * (noise_pred_text - noise_pred_uncond)
            
            latents = self.noise_scheduler.step(noise_pred, t, latents).prev_sample
        
        generated_image = self.decode_latents(latents)
        
        return generated_image
    
    def save_model(self, save_path: str):
        """모델 저장"""
        torch.save({
            'unet_state_dict': self.unet.state_dict(),
        }, save_path)
        print(f"모델 저장 완료: {save_path}")
    
    def load_model(self, load_path: str):
        """모델 로드"""
        checkpoint = torch.load(load_path, map_location=self.device, weights_only=True)
        self.unet.load_state_dict(checkpoint['unet_state_dict'])
        print(f"모델 로드 완료: {load_path}")


# ============================================================
# 3. Dataset 클래스
# ============================================================
class EmptyRoomDataset(Dataset):
    """가구 있는 방 -> 빈 방 변환 데이터셋"""
    
    def __init__(self, root_dir, transform=None):
        self.root_dir = Path(root_dir)
        self.transform = transform
        
        self.furnished_dir = self.root_dir / "furnished"
        self.empty_dir = self.root_dir / "empty"
        
        self.image_files = []
        
        if self.furnished_dir.exists() and self.empty_dir.exists():
            for img_file in sorted(self.furnished_dir.glob("*")):
                if img_file.suffix.lower() in ['.jpg', '.jpeg', '.png']:
                    empty_path = self.empty_dir / img_file.name
                    if empty_path.exists():
                        self.image_files.append(img_file.name)
        
        print(f"데이터셋에서 {len(self.image_files)}개의 이미지 쌍을 찾았습니다.")
        
        if self.transform is None:
            self.transform = transforms.Compose([
                transforms.Resize((512, 512)),
                transforms.ToTensor(),
                transforms.Normalize(mean=[0.5, 0.5, 0.5], 
                                   std=[0.5, 0.5, 0.5])
            ])
    
    def __len__(self):
        return len(self.image_files)
    
    def __getitem__(self, idx):
        img_name = self.image_files[idx]
        
        furnished_path = self.furnished_dir / img_name
        empty_path = self.empty_dir / img_name
        
        furnished_img = Image.open(furnished_path).convert('RGB')
        empty_img = Image.open(empty_path).convert('RGB')
        
        if self.transform:
            furnished_img = self.transform(furnished_img)
            empty_img = self.transform(empty_img)
        
        caption = "Remove all furniture and objects to create an empty room"
        
        return furnished_img, empty_img, caption


def create_dataloaders(root_dir, 
                       batch_size=4, 
                       num_workers=0,
                       train_split=0.9):
    """데이터로더 생성"""
    dataset = EmptyRoomDataset(root_dir)
    
    train_size = int(train_split * len(dataset))
    val_size = len(dataset) - train_size
    
    train_dataset, val_dataset = torch.utils.data.random_split(
        dataset, 
        [train_size, val_size]
    )
    
    print(f"학습 데이터: {len(train_dataset)}개")
    print(f"검증 데이터: {len(val_dataset)}개")
    
    train_loader = DataLoader(
        train_dataset,
        batch_size=batch_size,
        shuffle=True,
        num_workers=num_workers,
        pin_memory=True
    )
    
    val_loader = DataLoader(
        val_dataset,
        batch_size=batch_size,
        shuffle=False,
        num_workers=num_workers,
        pin_memory=True
    )
    
    return train_loader, val_loader


# ============================================================
# 4. Trainer 클래스
# ============================================================
class Trainer:
    def __init__(self,
                 model: EmptyRoomModel,
                 train_loader,
                 val_loader,
                 learning_rate: float = 1e-5,
                 num_epochs: int = 50,
                 save_dir: str = "checkpoints",
                 log_dir: str = "logs"):
        
        self.model = model
        self.train_loader = train_loader
        self.val_loader = val_loader
        self.num_epochs = num_epochs
        self.device = model.device
        
        # 옵티마이저
        self.optimizer = optim.AdamW(
            model.unet.parameters(),
            lr=learning_rate,
            weight_decay=0.01
        )
        
        # 스케줄러
        self.scheduler = optim.lr_scheduler.CosineAnnealingLR(
            self.optimizer,
            T_max=num_epochs
        )
        
        # 저장 경로
        self.save_dir = Path(save_dir)
        self.save_dir.mkdir(parents=True, exist_ok=True)
        
        # TensorBoard
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
        self.writer = SummaryWriter(f"{log_dir}/run_{timestamp}")
        
        # 최적 모델 추적
        self.best_val_loss = float('inf')
        self.global_step = 0
        
        # Mixed Precision Training
        self.scaler = torch.cuda.amp.GradScaler() if torch.cuda.is_available() else None
        
    def train_epoch(self, epoch):
        """한 에포크 학습"""
        self.model.train()
        total_loss = 0
        
        pbar = tqdm(self.train_loader, desc=f"Epoch {epoch+1}/{self.num_epochs}")
        
        for batch_idx, (furnished, empty, captions) in enumerate(pbar):
            furnished = furnished.to(self.device, non_blocking=True)
            empty = empty.to(self.device, non_blocking=True)
            
            # Mixed Precision Training
            if self.scaler is not None:
                with torch.cuda.amp.autocast():
                    loss = self.model(furnished, empty, captions)
                
                self.optimizer.zero_grad()
                self.scaler.scale(loss).backward()
                self.scaler.unscale_(self.optimizer)
                torch.nn.utils.clip_grad_norm_(self.model.unet.parameters(), 1.0)
                self.scaler.step(self.optimizer)
                self.scaler.update()
            else:
                loss = self.model(furnished, empty, captions)
                self.optimizer.zero_grad()
                loss.backward()
                torch.nn.utils.clip_grad_norm_(self.model.unet.parameters(), 1.0)
                self.optimizer.step()
            
            total_loss += loss.item()
            self.global_step += 1
            
            if batch_idx % 10 == 0:
                self.writer.add_scalar('Train/Loss', loss.item(), self.global_step)
                self.writer.add_scalar('Train/LR', self.optimizer.param_groups[0]['lr'], 
                                      self.global_step)
                
                if torch.cuda.is_available():
                    gpu_memory = torch.cuda.memory_allocated() / 1e9
                    self.writer.add_scalar('System/GPU_Memory_GB', gpu_memory, self.global_step)
            
            pbar.set_postfix({'loss': f'{loss.item():.4f}'})
        
        avg_loss = total_loss / len(self.train_loader)
        return avg_loss
    
    def validate(self, epoch):
        """검증"""
        self.model.eval()
        total_loss = 0
        
        with torch.no_grad():
            for furnished, empty, captions in tqdm(self.val_loader, desc="Validation"):
                furnished = furnished.to(self.device, non_blocking=True)
                empty = empty.to(self.device, non_blocking=True)
                
                if self.scaler is not None:
                    with torch.cuda.amp.autocast():
                        loss = self.model(furnished, empty, captions)
                else:
                    loss = self.model(furnished, empty, captions)
                    
                total_loss += loss.item()
        
        avg_loss = total_loss / len(self.val_loader)
        self.writer.add_scalar('Val/Loss', avg_loss, epoch)
        
        return avg_loss
    
    def save_checkpoint(self, epoch, val_loss, is_best=False):
        """체크포인트 저장"""
        checkpoint_path = self.save_dir / f"checkpoint_epoch_{epoch+1}.pt"
        self.model.save_model(str(checkpoint_path))
        
        if is_best:
            best_path = self.save_dir / "best_model.pt"
            self.model.save_model(str(best_path))
            print(f"✓ Best model saved! (Val Loss: {val_loss:.4f})")
    
    def train(self):
        """전체 학습 프로세스"""
        print("=" * 60)
        print("학습 시작")
        print("=" * 60)
        print(f"Device: {self.device}")
        print(f"Mixed Precision: {'Enabled' if self.scaler is not None else 'Disabled'}")
        print(f"Epochs: {self.num_epochs}")
        print(f"Train batches: {len(self.train_loader)}")
        print(f"Val batches: {len(self.val_loader)}")
        
        if torch.cuda.is_available():
            print(f"GPU: {torch.cuda.get_device_name(0)}")
            print(f"GPU Memory: {torch.cuda.get_device_properties(0).total_memory / 1e9:.2f} GB")
        
        print("=" * 60)
        
        start_time = time.time()
        
        for epoch in range(self.num_epochs):
            print(f"\nEpoch {epoch+1}/{self.num_epochs}")
            print("-" * 60)
            
            train_loss = self.train_epoch(epoch)
            print(f"Train Loss: {train_loss:.4f}")
            
            if torch.cuda.is_available():
                torch.cuda.empty_cache()
            
            val_loss = self.validate(epoch)
            print(f"Val Loss: {val_loss:.4f}")
            
            if torch.cuda.is_available():
                torch.cuda.empty_cache()
            
            self.scheduler.step()
            
            is_best = val_loss < self.best_val_loss
            if is_best:
                self.best_val_loss = val_loss
            
            if (epoch + 1) % 5 == 0 or is_best:
                self.save_checkpoint(epoch, val_loss, is_best)
        
        total_time = time.time() - start_time
        print("\n" + "=" * 60)
        print("학습 완료!")
        print(f"총 소요 시간: {total_time/3600:.2f} 시간")
        print(f"Best Val Loss: {self.best_val_loss:.4f}")
        print("=" * 60)
        
        self.writer.close()


# ============================================================
# 5. 메인 실행 코드
# ============================================================
print("=" * 60)
print("시스템 정보")
print("=" * 60)
print(f"PyTorch 버전: {torch.__version__}")
print(f"CUDA 사용 가능: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"CUDA 버전: {torch.version.cuda}")
    print(f"GPU 개수: {torch.cuda.device_count()}")
    print(f"현재 GPU: {torch.cuda.get_device_name(0)}")
    print(f"GPU 메모리: {torch.cuda.get_device_properties(0).total_memory / 1e9:.2f} GB")
print("=" * 60)
```

```python
# ============================================================
# 1. 필요한 패키지 import 및 설정
# ============================================================
import os
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from torch.utils.tensorboard import SummaryWriter
from pathlib import Path
from tqdm import tqdm
import time
from datetime import datetime
from PIL import Image
from torchvision import transforms

from diffusers import AutoencoderKL, UNet2DConditionModel, DDPMScheduler
from transformers import CLIPTextModel, CLIPTokenizer

# hf_transfer 비활성화
os.environ['HF_HUB_ENABLE_HF_TRANSFER'] = '0'

# ============================================================
# 2. EmptyRoomModel 클래스
# ============================================================
class EmptyRoomModel(nn.Module):
    """가구 제거를 위한 빈방 생성 모델"""
    
    def __init__(self, 
                 pretrained_model_name: str = "runwayml/stable-diffusion-v1-5",
                 device: str = "cuda" if torch.cuda.is_available() else "cpu"):
        super().__init__()
        
        self.device = device
        
        print(f"모델 로딩 중... (디바이스: {device})")
        print(f"사용 모델: {pretrained_model_name}")
        
        try:
            # VAE
            print("VAE 로딩 중...")
            self.vae = AutoencoderKL.from_pretrained(
                pretrained_model_name,
                subfolder="vae"
            ).to(device)
            if device == "cuda":
                self.vae = self.vae.half()
            
            # UNet
            print("UNet 로딩 중...")
            self.unet = UNet2DConditionModel.from_pretrained(
                pretrained_model_name,
                subfolder="unet"
            ).to(device)
            if device == "cuda":
                self.unet = self.unet.half()
            
            # Text Encoder
            print("Text Encoder 로딩 중...")
            self.text_encoder = CLIPTextModel.from_pretrained(
                pretrained_model_name,
                subfolder="text_encoder"
            ).to(device)
            if device == "cuda":
                self.text_encoder = self.text_encoder.half()
            
            # Tokenizer
            print("Tokenizer 로딩 중...")
            try:
                self.tokenizer = CLIPTokenizer.from_pretrained(
                    pretrained_model_name,
                    subfolder="tokenizer"
                )
            except:
                print("  -> 대체 tokenizer 사용 중...")
                from transformers import AutoTokenizer
                self.tokenizer = AutoTokenizer.from_pretrained(
                    "openai/clip-vit-large-patch14",
                    use_fast=False
                )
            
            # Noise Scheduler
            print("Scheduler 로딩 중...")
            self.noise_scheduler = DDPMScheduler.from_pretrained(
                pretrained_model_name,
                subfolder="scheduler"
            )
            
            # VAE scaling factor
            self.vae_scale_factor = 2 ** (len(self.vae.config.block_out_channels) - 1)
            
            # Freeze VAE and text encoder
            self.vae.requires_grad_(False)
            self.text_encoder.requires_grad_(False)
            
            print("모델 로딩 완료!")
            
        except Exception as e:
            print(f"모델 로딩 중 오류 발생: {e}")
            raise
    
    def encode_text(self, prompts):
        """텍스트를 임베딩으로 변환"""
        text_inputs = self.tokenizer(
            prompts,
            padding="max_length",
            max_length=self.tokenizer.model_max_length,
            truncation=True,
            return_tensors="pt"
        )
        
        text_input_ids = text_inputs.input_ids.to(self.device)
        
        with torch.no_grad():
            text_embeddings = self.text_encoder(text_input_ids)[0]
        
        return text_embeddings
    
    def encode_image(self, images):
        """이미지를 latent space로 인코딩"""
        with torch.no_grad():
            latents = self.vae.encode(images).latent_dist.sample()
            latents = latents * self.vae.config.scaling_factor
        return latents
    
    def decode_latents(self, latents):
        """latent를 이미지로 디코딩"""
        latents = latents / self.vae.config.scaling_factor
        with torch.no_grad():
            images = self.vae.decode(latents).sample
        return images
    
    def forward(self, 
                furnished_images: torch.Tensor,
                empty_images: torch.Tensor,
                captions: list):
        """학습을 위한 forward pass"""
        batch_size = furnished_images.shape[0]
        
        # 텍스트 인코딩
        text_embeddings = self.encode_text(captions)
        
        # 이미지를 latent로 변환
        furnished_latents = self.encode_image(furnished_images)
        empty_latents = self.encode_image(empty_images)
        
        # 노이즈 추가
        noise = torch.randn_like(empty_latents)
        timesteps = torch.randint(
            0, self.noise_scheduler.config.num_train_timesteps,
            (batch_size,), device=self.device
        ).long()
        
        noisy_latents = self.noise_scheduler.add_noise(empty_latents, noise, timesteps)
        
        # latent concat
        latent_model_input = torch.cat([noisy_latents, furnished_latents], dim=1)
        
        # 노이즈 예측
        noise_pred = self.unet(
            latent_model_input,
            timesteps,
            encoder_hidden_states=text_embeddings
        ).sample
        
        # Loss 계산
        loss = nn.functional.mse_loss(noise_pred, noise)
        
        return loss
    
    @torch.no_grad()
    def generate(self, 
                 furnished_image: torch.Tensor,
                 caption: str,
                 num_inference_steps: int = 50,
                 guidance_scale: float = 7.5):
        """추론: 가구가 있는 이미지에서 빈방 생성"""
        self.eval()
        
        text_embeddings = self.encode_text([caption])
        uncond_embeddings = self.encode_text([""])
        text_embeddings = torch.cat([uncond_embeddings, text_embeddings])
        
        furnished_latents = self.encode_image(furnished_image)
        latents = torch.randn_like(furnished_latents)
        
        self.noise_scheduler.set_timesteps(num_inference_steps)
        
        for t in self.noise_scheduler.timesteps:
            latent_model_input = torch.cat([latents] * 2)
            latent_model_input = torch.cat([latent_model_input, 
                                           furnished_latents.repeat(2, 1, 1, 1)], dim=1)
            
            noise_pred = self.unet(
                latent_model_input,
                t,
                encoder_hidden_states=text_embeddings
            ).sample
            
            noise_pred_uncond, noise_pred_text = noise_pred.chunk(2)
            noise_pred = noise_pred_uncond + guidance_scale * (noise_pred_text - noise_pred_uncond)
            
            latents = self.noise_scheduler.step(noise_pred, t, latents).prev_sample
        
        generated_image = self.decode_latents(latents)
        
        return generated_image
    
    def save_model(self, save_path: str):
        """모델 저장"""
        torch.save({
            'unet_state_dict': self.unet.state_dict(),
        }, save_path)
        print(f"모델 저장 완료: {save_path}")
    
    def load_model(self, load_path: str):
        """모델 로드"""
        checkpoint = torch.load(load_path, map_location=self.device, weights_only=True)
        self.unet.load_state_dict(checkpoint['unet_state_dict'])
        print(f"모델 로드 완료: {load_path}")


# ============================================================
# 3. Dataset 클래스
# ============================================================
class EmptyRoomDataset(Dataset):
    """가구 있는 방 -> 빈 방 변환 데이터셋"""
    
    def __init__(self, root_dir, transform=None):
        self.root_dir = Path(root_dir)
        self.transform = transform
        
        self.furnished_dir = self.root_dir / "furnished"
        self.empty_dir = self.root_dir / "empty"
        
        self.image_files = []
        
        if self.furnished_dir.exists() and self.empty_dir.exists():
            for img_file in sorted(self.furnished_dir.glob("*")):
                if img_file.suffix.lower() in ['.jpg', '.jpeg', '.png']:
                    empty_path = self.empty_dir / img_file.name
                    if empty_path.exists():
                        self.image_files.append(img_file.name)
        
        print(f"데이터셋에서 {len(self.image_files)}개의 이미지 쌍을 찾았습니다.")
        
        if self.transform is None:
            self.transform = transforms.Compose([
                transforms.Resize((512, 512)),
                transforms.ToTensor(),
                transforms.Normalize(mean=[0.5, 0.5, 0.5], 
                                   std=[0.5, 0.5, 0.5])
            ])
    
    def __len__(self):
        return len(self.image_files)
    
    def __getitem__(self, idx):
        img_name = self.image_files[idx]
        
        furnished_path = self.furnished_dir / img_name
        empty_path = self.empty_dir / img_name
        
        furnished_img = Image.open(furnished_path).convert('RGB')
        empty_img = Image.open(empty_path).convert('RGB')
        
        if self.transform:
            furnished_img = self.transform(furnished_img)
            empty_img = self.transform(empty_img)
        
        caption = "Remove all furniture and objects to create an empty room"
        
        return furnished_img, empty_img, caption


def create_dataloaders(root_dir, 
                       batch_size=4, 
                       num_workers=0,
                       train_split=0.9):
    """데이터로더 생성"""
    dataset = EmptyRoomDataset(root_dir)
    
    # 데이터셋이 비어있는지 확인
    if len(dataset) == 0:
        print("\n" + "=" * 60)
        print("❌ 오류: 데이터셋이 비어있습니다!")
        print("=" * 60)
        print(f"확인한 경로: {root_dir}")
        print(f"필요한 폴더 구조:")
        print(f"  {root_dir}/")
        print(f"    ├── furnished/  (가구 있는 이미지)")
        print(f"    └── empty/      (빈 방 이미지)")
        print("\n해결 방법:")
        print("1. 데이터셋 경로가 올바른지 확인")
        print("2. furnished와 empty 폴더가 존재하는지 확인")
        print("3. 각 폴더에 이미지 파일(.jpg, .png)이 있는지 확인")
        print("4. 파일 이름이 양쪽 폴더에서 일치하는지 확인")
        
        # 실제 폴더 확인
        root_path = Path(root_dir)
        if root_path.exists():
            print(f"\n실제 폴더 내용:")
            for item in root_path.iterdir():
                print(f"  - {item.name}")
        else:
            print(f"\n경로가 존재하지 않습니다: {root_dir}")
        
        raise ValueError("데이터셋이 비어있습니다. 위의 해결 방법을 확인하세요.")
    
    train_size = int(train_split * len(dataset))
    val_size = len(dataset) - train_size
    
    # 최소 1개씩은 있어야 함
    if train_size == 0:
        train_size = 1
        val_size = len(dataset) - 1
    if val_size == 0:
        val_size = 1
        train_size = len(dataset) - 1
    
    train_dataset, val_dataset = torch.utils.data.random_split(
        dataset, 
        [train_size, val_size]
    )
    
    print(f"학습 데이터: {len(train_dataset)}개")
    print(f"검증 데이터: {len(val_dataset)}개")
    
    train_loader = DataLoader(
        train_dataset,
        batch_size=batch_size,
        shuffle=True,
        num_workers=num_workers,
        pin_memory=True
    )
    
    val_loader = DataLoader(
        val_dataset,
        batch_size=batch_size,
        shuffle=False,
        num_workers=num_workers,
        pin_memory=True
    )
    
    return train_loader, val_loader


# ============================================================
# 4. Trainer 클래스
# ============================================================
class Trainer:
    def __init__(self,
                 model: EmptyRoomModel,
                 train_loader,
                 val_loader,
                 learning_rate: float = 1e-5,
                 num_epochs: int = 50,
                 save_dir: str = "checkpoints",
                 log_dir: str = "logs"):
        
        self.model = model
        self.train_loader = train_loader
        self.val_loader = val_loader
        self.num_epochs = num_epochs
        self.device = model.device
        
        # 옵티마이저
        self.optimizer = optim.AdamW(
            model.unet.parameters(),
            lr=learning_rate,
            weight_decay=0.01
        )
        
        # 스케줄러
        self.scheduler = optim.lr_scheduler.CosineAnnealingLR(
            self.optimizer,
            T_max=num_epochs
        )
        
        # 저장 경로
        self.save_dir = Path(save_dir)
        self.save_dir.mkdir(parents=True, exist_ok=True)
        
        # TensorBoard
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
        self.writer = SummaryWriter(f"{log_dir}/run_{timestamp}")
        
        # 최적 모델 추적
        self.best_val_loss = float('inf')
        self.global_step = 0
        
        # Mixed Precision Training
        self.scaler = torch.cuda.amp.GradScaler() if torch.cuda.is_available() else None
        
    def train_epoch(self, epoch):
        """한 에포크 학습"""
        self.model.train()
        total_loss = 0
        
        pbar = tqdm(self.train_loader, desc=f"Epoch {epoch+1}/{self.num_epochs}")
        
        for batch_idx, (furnished, empty, captions) in enumerate(pbar):
            furnished = furnished.to(self.device, non_blocking=True)
            empty = empty.to(self.device, non_blocking=True)
            
            # Mixed Precision Training
            if self.scaler is not None:
                with torch.cuda.amp.autocast():
                    loss = self.model(furnished, empty, captions)
                
                self.optimizer.zero_grad()
                self.scaler.scale(loss).backward()
                self.scaler.unscale_(self.optimizer)
                torch.nn.utils.clip_grad_norm_(self.model.unet.parameters(), 1.0)
                self.scaler.step(self.optimizer)
                self.scaler.update()
            else:
                loss = self.model(furnished, empty, captions)
                self.optimizer.zero_grad()
                loss.backward()
                torch.nn.utils.clip_grad_norm_(self.model.unet.parameters(), 1.0)
                self.optimizer.step()
            
            total_loss += loss.item()
            self.global_step += 1
            
            if batch_idx % 10 == 0:
                self.writer.add_scalar('Train/Loss', loss.item(), self.global_step)
                self.writer.add_scalar('Train/LR', self.optimizer.param_groups[0]['lr'], 
                                      self.global_step)
                
                if torch.cuda.is_available():
                    gpu_memory = torch.cuda.memory_allocated() / 1e9
                    self.writer.add_scalar('System/GPU_Memory_GB', gpu_memory, self.global_step)
            
            pbar.set_postfix({'loss': f'{loss.item():.4f}'})
        
        avg_loss = total_loss / len(self.train_loader)
        return avg_loss
    
    def validate(self, epoch):
        """검증"""
        self.model.eval()
        total_loss = 0
        
        with torch.no_grad():
            for furnished, empty, captions in tqdm(self.val_loader, desc="Validation"):
                furnished = furnished.to(self.device, non_blocking=True)
                empty = empty.to(self.device, non_blocking=True)
                
                if self.scaler is not None:
                    with torch.cuda.amp.autocast():
                        loss = self.model(furnished, empty, captions)
                else:
                    loss = self.model(furnished, empty, captions)
                    
                total_loss += loss.item()
        
        avg_loss = total_loss / len(self.val_loader)
        self.writer.add_scalar('Val/Loss', avg_loss, epoch)
        
        return avg_loss
    
    def save_checkpoint(self, epoch, val_loss, is_best=False):
        """체크포인트 저장"""
        checkpoint_path = self.save_dir / f"checkpoint_epoch_{epoch+1}.pt"
        self.model.save_model(str(checkpoint_path))
        
        if is_best:
            best_path = self.save_dir / "best_model.pt"
            self.model.save_model(str(best_path))
            print(f"✓ Best model saved! (Val Loss: {val_loss:.4f})")
    
    def train(self):
        """전체 학습 프로세스"""
        print("=" * 60)
        print("학습 시작")
        print("=" * 60)
        print(f"Device: {self.device}")
        print(f"Mixed Precision: {'Enabled' if self.scaler is not None else 'Disabled'}")
        print(f"Epochs: {self.num_epochs}")
        print(f"Train batches: {len(self.train_loader)}")
        print(f"Val batches: {len(self.val_loader)}")
        
        if torch.cuda.is_available():
            print(f"GPU: {torch.cuda.get_device_name(0)}")
            print(f"GPU Memory: {torch.cuda.get_device_properties(0).total_memory / 1e9:.2f} GB")
        
        print("=" * 60)
        
        start_time = time.time()
        
        for epoch in range(self.num_epochs):
            print(f"\nEpoch {epoch+1}/{self.num_epochs}")
            print("-" * 60)
            
            train_loss = self.train_epoch(epoch)
            print(f"Train Loss: {train_loss:.4f}")
            
            if torch.cuda.is_available():
                torch.cuda.empty_cache()
            
            val_loss = self.validate(epoch)
            print(f"Val Loss: {val_loss:.4f}")
            
            if torch.cuda.is_available():
                torch.cuda.empty_cache()
            
            self.scheduler.step()
            
            is_best = val_loss < self.best_val_loss
            if is_best:
                self.best_val_loss = val_loss
            
            if (epoch + 1) % 5 == 0 or is_best:
                self.save_checkpoint(epoch, val_loss, is_best)
        
        total_time = time.time() - start_time
        print("\n" + "=" * 60)
        print("학습 완료!")
        print(f"총 소요 시간: {total_time/3600:.2f} 시간")
        print(f"Best Val Loss: {self.best_val_loss:.4f}")
        print("=" * 60)
        
        self.writer.close()


# ============================================================
# 5. 메인 실행 코드
# ============================================================
print("=" * 60)
print("시스템 정보")
print("=" * 60)
print(f"PyTorch 버전: {torch.__version__}")
print(f"CUDA 사용 가능: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"CUDA 버전: {torch.version.cuda}")
    print(f"GPU 개수: {torch.cuda.device_count()}")
    print(f"현재 GPU: {torch.cuda.get_device_name(0)}")
    print(f"GPU 메모리: {torch.cuda.get_device_properties(0).total_memory / 1e9:.2f} GB")
print("=" * 60)
```

```python
import torch
import numpy as np
from skimage.metrics import structural_similarity as ssim
from skimage.metrics import peak_signal_noise_ratio as psnr
import lpips
import clip
from PIL import Image
from typing import Dict


class EvaluationMetrics:
    """
    이미지 생성 품질 평가 지표
    - SSIM: Structural Similarity Index
    - LPIPS: Learned Perceptual Image Patch Similarity
    - PSNR: Peak Signal-to-Noise Ratio
    - CLIPScore: CLIP-based semantic similarity
    """
    def __init__(self, device: str = "cuda" if torch.cuda.is_available() else "cpu"):
        self.device = device
        
        # LPIPS 모델 로드
        print("LPIPS 모델 로딩 중...")
        self.lpips_model = lpips.LPIPS(net='alex').to(device)
        
        # CLIP 모델 로드
        print("CLIP 모델 로딩 중...")
        self.clip_model, self.clip_preprocess = clip.load("ViT-B/32", device=device)
        self.clip_model = self.clip_model.to(device)
        
        print("평가 지표 준비 완료!")
    
    def compute_ssim(self, img1: np.ndarray, img2: np.ndarray) -> float:
        """SSIM 계산 (구조적 유사도, 높을수록 좋음 0~1)"""
        if len(img1.shape) == 3:
            img1_gray = np.mean(img1, axis=2).astype(np.uint8)
            img2_gray = np.mean(img2, axis=2).astype(np.uint8)
        else:
            img1_gray = img1
            img2_gray = img2
        score = ssim(img1_gray, img2_gray, data_range=255)
        return float(score)
    
    def compute_psnr(self, img1: np.ndarray, img2: np.ndarray) -> float:
        """PSNR 계산 (픽셀 레벨 유사도, 높을수록 좋음 20~50)"""
        score = psnr(img1, img2, data_range=255)
        return float(score)
    
    def compute_lpips(self, img1: torch.Tensor, img2: torch.Tensor) -> float:
        """LPIPS 계산 (지각적 유사도, 낮을수록 좋음 0~1)"""
        # ✅ 디바이스 통일
        img1 = img1.to(self.device)
        img2 = img2.to(self.device)
        self.lpips_model = self.lpips_model.to(self.device)

        with torch.no_grad():
            score = self.lpips_model(img1, img2)
        return float(score.item())
    
    def compute_clip_score(self, image: Image.Image, text: str) -> float:
        """CLIPScore 계산 (이미지-텍스트 유사도, 높을수록 좋음 0~100)"""
        image_input = self.clip_preprocess(image).unsqueeze(0).to(self.device)
        text_input = clip.tokenize([text]).to(self.device)

        with torch.no_grad():
            image_features = self.clip_model.encode_image(image_input)
            text_features = self.clip_model.encode_text(text_input)

            # 정규화
            image_features = image_features / image_features.norm(dim=-1, keepdim=True)
            text_features = text_features / text_features.norm(dim=-1, keepdim=True)

            similarity = (image_features @ text_features.T).squeeze()

        return float(similarity.item() * 100)
    
    def evaluate_pair(self, 
                      generated_img: torch.Tensor,
                      target_img: torch.Tensor,
                      generated_pil: Image.Image,
                      caption: str) -> Dict[str, float]:
        """생성된 이미지와 타겟 이미지를 모든 지표로 평가"""
        # Tensor -> numpy 변환 [0, 255]
        gen_np = self._tensor_to_numpy(generated_img)
        target_np = self._tensor_to_numpy(target_img)
        
        metrics = {
            'ssim': self.compute_ssim(gen_np, target_np),
            'psnr': self.compute_psnr(gen_np, target_np),
            'lpips': self.compute_lpips(generated_img, target_img),
            'clip_score': self.compute_clip_score(generated_pil, caption)
        }
        
        return metrics
    
    def _tensor_to_numpy(self, tensor: torch.Tensor) -> np.ndarray:
        """Tensor [-1, 1] → Numpy [0, 255]"""
        img = tensor.squeeze(0).permute(1, 2, 0).cpu().numpy()
        img = ((img + 1) / 2 * 255).clip(0, 255).astype(np.uint8)
        return img
    
    def _numpy_to_tensor(self, img: np.ndarray) -> torch.Tensor:
        """Numpy [0, 255] → Tensor [-1, 1]"""
        img = torch.from_numpy(img).permute(2, 0, 1).float()
        img = (img / 255.0) * 2 - 1
        return img.unsqueeze(0)


def print_metrics(metrics: Dict[str, float]):
    """평가 지표 출력"""
    print("\n평가 지표:")
    print("-" * 40)
    print(f"SSIM:       {metrics['ssim']:.4f}  (↑ 높을수록 좋음, 0~1)")
    print(f"PSNR:       {metrics['psnr']:.4f} dB  (↑ 높을수록 좋음)")
    print(f"LPIPS:      {metrics['lpips']:.4f}  (↓ 낮을수록 좋음, 0~1)")
    print(f"CLIPScore:  {metrics['clip_score']:.2f}  (↑ 높을수록 좋음, 0~100)")
    print("-" * 40)


if __name__ == "__main__":
    print("평가 지표 모듈 테스트...")
    evaluator = EvaluationMetrics()

    # 더미 데이터 생성
    img1 = torch.randn(1, 3, 512, 512)
    img2 = torch.randn(1, 3, 512, 512)

    # numpy 변환 테스트
    img1_np = evaluator._tensor_to_numpy(img1)
    print(f"Numpy image shape: {img1_np.shape}, dtype: {img1_np.dtype}")

    # PIL 이미지 생성
    img1_pil = Image.fromarray(img1_np)

    # 평가 테스트
    metrics = evaluator.evaluate_pair(
        img1, img2, img1_pil, "A modern empty room with white walls"
    )

    print_metrics(metrics)
    print("\n평가 지표 모듈 테스트 완료!")
```

```python
import torch
import numpy as np
from pathlib import Path
from PIL import Image
from tqdm import tqdm
import json
import matplotlib.pyplot as plt

from model import EmptyRoomModel
from dataset_loader import create_dataloaders
from metrics import EvaluationMetrics, print_metrics


class Tester:
    def __init__(self, 
                 model: EmptyRoomModel,
                 test_loader,
                 test_dataset,
                 save_dir: str = "test_results"):
        self.model = model
        self.test_loader = test_loader
        self.test_dataset = test_dataset
        self.device = model.device
        
        self.save_dir = Path(save_dir)
        self.save_dir.mkdir(parents=True, exist_ok=True)
        
        # 평가 지표 초기화
        self.evaluator = EvaluationMetrics(device=self.device)
    
    def test_model(self, num_inference_steps: int = 50):
        """
        테스트 데이터셋 전체에 대해 평가
        """
        self.model.eval()
        
        all_metrics = {
            'ssim': [],
            'psnr': [],
            'lpips': [],
            'clip_score': []
        }
        
        print("=" * 60)
        print("모델 테스트 시작")
        print("=" * 60)
        
        for idx in tqdm(range(len(self.test_dataset)), desc="Testing"):
            # 원본 이미지 로드
            furnished_pil, target_pil, caption = self.test_dataset.get_raw_images(idx)
            
            # 전처리
            furnished_tensor = self.test_dataset.transform(furnished_pil).unsqueeze(0).to(self.device)
            target_tensor = self.test_dataset.transform(target_pil).unsqueeze(0).to(self.device)
            
            # 이미지 생성
            generated_tensor = self.model.generate(
                furnished_tensor,
                caption,
                num_inference_steps=num_inference_steps
            )
            
            # Tensor를 PIL로 변환
            generated_pil = self._tensor_to_pil(generated_tensor)
            
            # 평가
            metrics = self.evaluator.evaluate_pair(
                generated_tensor,
                target_tensor,
                generated_pil,
                caption
            )
            
            # 메트릭 저장
            for key in all_metrics:
                all_metrics[key].append(metrics[key])
            
            # 시각화 (처음 10개만)
            if idx < 10:
                self._save_comparison(
                    idx,
                    furnished_pil,
                    target_pil,
                    generated_pil,
                    caption,
                    metrics
                )
        
        # 평균 계산
        avg_metrics = {
            key: np.mean(values) for key, values in all_metrics.items()
        }
        
        # 결과 출력
        print("\n" + "=" * 60)
        print("테스트 결과 (평균)")
        print("=" * 60)
        print_metrics(avg_metrics)
        
        # 결과 저장
        self._save_results(all_metrics, avg_metrics)
        
        return avg_metrics
    
    def _tensor_to_pil(self, tensor: torch.Tensor) -> Image.Image:
        """Tensor를 PIL Image로 변환"""
        img = tensor.squeeze(0).permute(1, 2, 0).cpu().numpy()
        img = ((img + 1) / 2 * 255).clip(0, 255).astype(np.uint8)
        return Image.fromarray(img)
    
    def _save_comparison(self, 
                        idx: int,
                        furnished: Image.Image,
                        target: Image.Image,
                        generated: Image.Image,
                        caption: str,
                        metrics: dict):
        """비교 이미지 저장"""
        fig, axes = plt.subplots(1, 3, figsize=(15, 5))
        
        axes[0].imshow(furnished)
        axes[0].set_title("Input (Furnished)")
        axes[0].axis('off')
        
        axes[1].imshow(target)
        axes[1].set_title("Target (Empty)")
        axes[1].axis('off')
        
        axes[2].imshow(generated)
        axes[2].set_title("Generated (Empty)")
        axes[2].axis('off')
        
        # 메트릭 정보 추가
        metric_text = (f"SSIM: {metrics['ssim']:.3f} | "
                      f"PSNR: {metrics['psnr']:.2f} | "
                      f"LPIPS: {metrics['lpips']:.3f} | "
                      f"CLIP: {metrics['clip_score']:.1f}")
        
        plt.suptitle(f"Test Sample {idx}\nCaption: {caption[:60]}...\n{metric_text}", 
                    fontsize=10)
        plt.tight_layout()
        
        save_path = self.save_dir / f"comparison_{idx}.png"
        plt.savefig(save_path, dpi=150, bbox_inches='tight')
        plt.close()
    
    def _save_results(self, all_metrics: dict, avg_metrics: dict):
        """결과를 JSON으로 저장"""
        results = {
            'average': avg_metrics,
            'std': {
                key: float(np.std(values)) for key, values in all_metrics.items()
            },
            'individual': {
                idx: {key: all_metrics[key][idx] for key in all_metrics}
                for idx in range(len(all_metrics['ssim']))
            }
        }
        
        results_file = self.save_dir / "test_results.json"
        with open(results_file, 'w', encoding='utf-8') as f:
            json.dump(results, f, indent=2)
        
        print(f"\n결과 저장 완료: {results_file}")


def generate_from_gemini_image(model: EmptyRoomModel, 
                               image_path: str,
                               caption: str = "A modern empty room",
                               save_path: str = "gemini_test_result.png"):
    """
    Gemini로 생성한 인테리어 이미지에서 가구 제거 테스트
    
    Args:
        model: 학습된 모델
        image_path: Gemini로 생성한 인테리어 이미지 경로
        caption: 방 설명
        save_path: 결과 저장 경로
    """
    print("\n" + "=" * 60)
    print("Gemini 이미지 테스트")
    print("=" * 60)
    
    # 이미지 로드
    image = Image.open(image_path).convert('RGB')
    
    # 전처리
    from torchvision import transforms
    transform = transforms.Compose([
        transforms.Resize((512, 512)),
        transforms.ToTensor(),
        transforms.Normalize([0.5], [0.5])
    ])
    
    image_tensor = transform(image).unsqueeze(0).to(model.device)
    
    # 생성
    print(f"이미지 생성 중... (Caption: {caption})")
    with torch.no_grad():
        generated_tensor = model.generate(image_tensor, caption)
    
    # 결과 저장
    generated_img = model.decode_latents(generated_tensor)
    generated_pil = transforms.ToPILImage()(
        (generated_img.squeeze(0).cpu() + 1) / 2
    )
    
    # 비교 이미지 생성
    fig, axes = plt.subplots(1, 2, figsize=(12, 6))
    
    axes[0].imshow(image)
    axes[0].set_title("Input (Furnished Room)")
    axes[0].axis('off')
    
    axes[1].imshow(generated_pil)
    axes[1].set_title("Generated (Empty Room)")
    axes[1].axis('off')
    
    plt.suptitle(f"Gemini Image Test\nCaption: {caption}")
    plt.tight_layout()
    plt.savefig(save_path, dpi=150, bbox_inches='tight')
    plt.close()
    
    print(f"결과 저장 완료: {save_path}")
    
    return generated_pil


def main():
    BASE_PATH = os.getcwd()  # 현재 디렉토리
    DATASET_PATH = os.path.join(BASE_PATH, "dataset")
    MODEL_PATH = os.path.join(BASE_PATH, "checkpoints", "best_model.pt")
    RESULTS_PATH = os.path.join(BASE_PATH, "test_results")
    
    # 데이터로더 생성
    print("데이터로더 생성 중...")
    _, _, test_loader, _, _, test_dataset = create_dataloaders(
        dataset_path=DATASET_PATH,
        batch_size=1,
        num_workers=2
    )
    
    # 모델 로드
    print("\n모델 로드 중...")
    model = EmptyRoomModel()
    model.load_model(MODEL_PATH)
    model.eval()
    
    # 테스트 실행
    tester = Tester(
        model=model,
        test_loader=test_loader,
        test_dataset=test_dataset,
        save_dir=RESULTS_PATH
    )
    
    avg_metrics = tester.test_model(num_inference_steps=50)
    
    # Gemini 이미지 테스트 (이미지가 있는 경우)
    gemini_image_path = os.path.join(BASE_PATH, "gemini_test_image.jpg")
    if Path(gemini_image_path).exists():
        print("\nGemini 생성 이미지 테스트...")
        generate_from_gemini_image(
            model=model,
            image_path=gemini_image_path,
            caption="A modern empty room with clean walls",
            save_path=os.path.join(RESULTS_PATH, "gemini_test_result.png")
        )
    else:
        print(f"\nGemini 테스트 이미지를 찾을 수 없습니다: {gemini_image_path}")
        print("Gemini로 인테리어 이미지를 생성하여 해당 경로에 저장하세요.")


if __name__ == "__main__":
    main()
```
